```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plot 
import tensorflow as tf 
from tensorflow.keras.models import load_model
from tensorflow.keras.datasets import imdb 
from tensorflow.keras.utils import pad_sequences
```

```python
words_by_index = imdb.get_word_index()
reversed_words_index = {value:key for key,value in words_by_index.items()}
```

```python
def decode(review_code) : 
    return " ".join([words_by_index.get(word,"?")-3 for word in review_code])
```

```python
def encode(review) : 
    review = review.lower().split()
    encode_review = [reversed_words_index.get(word,2)+3 for word in review]
    seq = pad_sequences([encode_review],maxlen=500,padding="post")
    return seq
```

```python
model = load_model("my_model.h5")
```

```python
def prediction(text) : 
    preprocessed = encode(text)
    prediction = model.predict(preprocessed)
    return "positive" if prediction > 0.5 else "negative"
```

```python
text = "i love that food but joe said that was terrible"
```

```python
result = prediction(text)
```
